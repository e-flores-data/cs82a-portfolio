```python
import pandas as pd
```

```python
cust = pd.read_csv("custdata.tsv", sep="\t")

print(cust.shape)
display(cust[["age", "income", "num.vehicles"]].describe())
```

```python
cust[["age", "income", "num.vehicles"]].corr()
```

Correlation before cleaning data

```python
print("Age 0 or less:",(cust["age"]<= 0).sum())
```

```python
print("Age over 100:",(cust["age"] > 100).sum())
```

```python
print("Negative income:",(cust["income"] < 0).sum())
```

```python
print("Missing vehicle values:",cust["num.vehicles"].isna().sum())
```

```python
cust_clean = cust[cust["age"] > 0]
cust_clean = cust_clean[cust_clean["age"] <= 100]
cust_clean = cust_clean[cust_clean["income"] >= 0]
cust_clean = cust_clean.dropna(subset=["num.vehicles"])
```

```python
print(cust_clean.shape)
print("Total rows removed:", len(cust) - len(cust_clean))
```

Correlations after cleaning data

```python
cust_clean[["age", "income", "num.vehicles"]].corr()
```

Cleaning data: removed 3 ages that were 0 or less, removed 8 ages that were over 100, removed negative income, removed 56 rows that had missing vehicle data. A total of 68 rows were removed.There were all removed because they seemed like errors (invalid) or simply missing data and removed to not affect any correlations.
