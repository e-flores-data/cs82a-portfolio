```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
cust = pd.read_csv("custdata.tsv", sep="\t")
print(cust.shape)
display(cust.head())
print(cust.columns)
```

```python
cust["housing.type"].isna().sum()
```

```python
cust_clean = cust.dropna(subset=["housing.type"])
```

```python
print(cust_clean.shape)
print(cust_clean["housing.type"].isna().sum())
```

```python
cust_clean["housing.type"].value_counts()
```

```python
housing_order = cust_clean["housing.type"].value_counts().index
sns.countplot(data=cust_clean, x="housing.type", order=housing_order)
plt.title("how many customers fall into each housing type.")
plt.xlabel("Housing Type")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.show()
```
