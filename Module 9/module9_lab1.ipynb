```python
def pounds_to_kilograms(pounds):
    return pounds * 0.453592
```

```python
print(pounds_to_kilograms(1))
print(pounds_to_kilograms(100))
print(pounds_to_kilograms(180))
```

```python
print("1 pound in kilograms is", pounds_to_kilograms(1))
print("100 pounds in kilograms is", pounds_to_kilograms(100))
print("My weight in kilograms equals", pounds_to_kilograms(180))
```
