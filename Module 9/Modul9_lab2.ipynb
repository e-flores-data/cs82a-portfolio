Migration: Google colab to Jupyter **notebook**

I migrated my Module 8 Lab 1 notebook from Google Colab to Jupyter Notebook. I chose this file because it contains Python code, markdown text, data analysis, and charts. The code, layout, and text were expected to transfer without issues because Google Colab and Jupyter Notebook both use Python and the .ipynb notebook format.

The first step was downloading the .ipynb file from Google Colab. Since I already had Miniforge and Jupyter running, I did not need to reopen them. I uploaded the downloaded Module 8 Lab 1 notebook into Jupyter, where it opened as a new notebook. The Python code, markdown text, headings, layout, and previously saved outputs appeared to match the version in Google Colab.

Although the notebook transferred, its separate data file did not transfer with it. When I restarted the kernel and ran all cells, I received a FileNotFoundError because Jupyter could not locate penguins.csv. I uploaded the CSV file into the same Jupyter folder as the notebook and restarted and ran all cells again. After the CSV was added, the code and charts ran without errors and produced the same results.

The two platforms use different storage locations, so notebooks and their related data files must be moved separately. Colab uses a Google account and cloud storage, while my local Jupyter environment uses Miniforge and files stored on my computer. My required packages were already installed in my cs82a environment. If a notebook used a package that was not installed in the new environment, that package would need to be installed before the code could run.

The main risk is assuming that downloading the notebook automatically includes its data files. A notebook may open normally and display saved results, even when the original CSV is missing. However, the code will fail when someone restarts the notebook and runs all cells. The data itself may not be lost, but the analysis cannot be reproduced until the correct file is located, uploaded, and placed where the code expects to find it. To prevent this problem, I would move the notebook and all required data files together, confirm the filenames and locations, and restart and run all cells after the migration.
