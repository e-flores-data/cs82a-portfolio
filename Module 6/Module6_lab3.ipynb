```python
import sqlite3
import pandas as pd
con = sqlite3.connect("customers.db")
```

```python
con.execute("DROP TABLE IF EXISTS customers")
con.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    total_spent REAL)
""")
```

```python
pd.read_sql("SELECT * FROM customers", con)
```

```python
pd.read_sql("""
SELECT *
FROM customers
WHERE total_spent > 150
""", con)
```

```python
pd.read_sql("""
SELECT *
FROM customers
ORDER BY total_spent DESC
""", con)
```

```python
pd.read_sql("""
SELECT *
FROM customers
WHERE name LIKE '%an%'
""", con)
```

```python
import matplotlib.pyplot as plt
chart_data = pd.read_sql("""
SELECT name, total_spent
FROM customers
ORDER BY total_spent DESC
""", con)

plt.bar(chart_data["name"], chart_data["total_spent"])
plt.title("Angel Potter Has the Highest Total Spending")
plt.xlabel("Customer Name")
plt.ylabel("Total Spent (Dollars)")
plt.xticks(rotation=45)
plt.show()
```
