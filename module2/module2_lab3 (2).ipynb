cutoffs:
- 20°F and below: Freezing
- 21-55°F: Cold
- 56- 70°F: Mild
- 71- 85°F: Warm
- 85°F and higher: Hot

```python
temperatures = [50, 61, 73, 82, 57, 72, 86, 70,85,55]
```

```python
for temperature in temperatures:
    if temperature<= 20:
        label="Freezing"
    elif temperature<= 55:
        label="Cold"
    elif temperature<= 70:
        label="Mild"
    elif temperature <= 85:
        label="warm"
    else:
        label = "Hot"
    print(f"{temperature}°F: {label}")

    
```

I added 70, 85 and 55 to test the boundaries after the initial run.
