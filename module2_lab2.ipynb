```python
def multiply_for(number,times):
    result = 0
    for count in range(times):
        result=result+number
    return result
```

```python
for_result = multiply_for(7,6)
print("For loop result:", for_result)
```

```python
def multiply_while(number,times):
    result=0
    count=0
    while count<times:
        result=result+number
        count=count+1
    return result
```

```python
while_result=multiply_while(7,6)
print("While loop result:",while_result)
```

```python
print("For loop:",multiply_for(7,6))
print("While loop:", multiply_while(7,6))
print("Real product:",7*6)
```

```python
print("For loop:", multiply_for(5,8))
print("While loop:", multiply_while(5,8))
print("Real product:",5*8)
```

```python
print("For loop:", multiply_for(20,4))
print("While loop:", multiply_while(20,4))
print("Real product:",20*4)
```

The while loop terminates once the count reaches the number of times. So, in the while loop for (7, 6), it will repeat 7 until the count reaches 6. So what we cannot see is the loop going like: 7 + 7 + 7 + 7 + 7 + 7 = 42.
