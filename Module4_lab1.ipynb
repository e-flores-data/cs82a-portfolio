```python
import pandas as pd

df = pd.read_csv("sales_clean.csv")
```

```python
print(df.shape)
display(df.head())
```

```python
print(df["price"].mean())
print(df["price"].median())
print(df["price"].std())
```

```python
print(df["qty"].mean())
print(df["qty"].median())
print(df["qty"].std())
```

When looking at reporting the price between the mean and median, I would choose the median. I would make this choice because when looking at the data it seemed I had less than 20 high prices that would skew the data and create a higher mean, which would not be as truly representative of the average price. The median adjusts for the outliers so based on this data set, it appears to be more reliable of an average to report for the price.

```python
ax = df["price"].plot(kind="hist", bins=20, title="Distribution of Price")
```

The histogram is right skewed and shows that my earlier observation that there seemed to be less than 20 high prices is accurate. This supports using the median over mean.

```python
df["revenue"]=df["price"]*df["qty"]
```

```python
display(df.head())
```

```python
df.corr(numeric_only=True)
```

The strongest correlation is between price and revenue at .817699

```python
ax= df.plot.scatter(x="qty",y="revenue", title="Quantity vs. Revenue")
```

Analysis: The relationship shows that quanity and revenue increase together but it is not a strong correlation. The correlation is .334196 which is weak compared to a perfect correlation of 1. One caution with this data is that price is a confounder. So that while quanity affects revenue, price also affects the revenue. There were items listed at high prices for a single item and produce more revenue than other items combined. We also see evidence for this because the correlation between price and revenue is .817699 and closer to 1, making is a stronger correlation.
