```python
ax = df.plot.scatter(
    x="X",
    y="Y",
    title="Higher List Prices Have Higher Best Prices")
ax.set_xlabel("List Price (Thousands of Dollars)")
ax.set_ylabel("Best Price (Thousands of Dollars)")
```

Prediction- A truck with a list price of 18,000 will have a predicted best price of around 15,750. There is a strong correlation of r= 0.99854 which leads me to be confident in this assumption as it is close to 1 and a near straight line on the scatter plot. The data range may not include a wide range of prices which can be a limitation when predicting prices that fall outside of this dataset.
