```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")

print(df.shape)
display(df.head(10))
print(df.dtypes)
```

Observations: 3 problems I can see with this data
1. Date, not consistent formatting.
2. Price column missing data.
3. Zip codes are missing numbers (0).

```python
df.isna().sum()
```

```python
fill_value = df["price"].median()
df["price"] = df["price"].fillna(fill_value)
```

```python
print("Median price used:", fill_value)
df.isna().sum()
```

```python
df.isna().sum()
```

```python
print(df.duplicated().sum())
df = df.drop_duplicates()
```

```python
print("Shape after removing duplicates:", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(4)
```

```python
print(df["zip"].head(10))
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)
```

```python
print(df["zip"].head(10))
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")
```

```python
df.to_csv("sales_clean.csv", index=False)
```

```python
Cleaning log: 
-Started with 300 rows.
-There were 12 missing prices identified, used the median price to fill.
-Identified 8 duplicated rows, removed the duplicated rows.
-Zip codes were missing the zeros, added zeros to zip codes.
-Some negative quantities identified, I chose to treat them as returns (returns to flag). I chose this option because when reviewing the data it appears the negative quanities relate to an order placed earlier. 
For example I saw that order 1250 ordered April 30, 2026 ordered 5 desk lamps with a zip code of 10001 followed by a order 1233 on May 09,2026 showing -4 quanitity from the same zip code of 10001.
With this finding and without more information I feel that the orders are related and to be kept in the data with the added return flag for enteries.
A wrong choice can mislead a business decision on how many of what items to order, as it can cause issues in inventory. 

```
